```python
import os
import requests
from dotenv import load_dotenv

# Load environment variables
load_dotenv()

TMDB_API_KEY = os.getenv("TMDB_API_KEY") or os.getenv("TMDB_key")
TMDB_BASE = "https://api.themoviedb.org/3"

def get_streaming_providers(series_name: str, country_code: str = "US"):
    """
    Searches for a TV series on TMDB and retrieves its streaming providers 
    for a specific country (defaulting to 'US').
    """
    if not TMDB_API_KEY:
        raise ValueError("TMDB API Key not found. Please ensure it is set in your environment or .env file.")

    headers = {
        "accept": "application/json",
        "Authorization": f"Bearer {TMDB_API_KEY}" if len(TMDB_API_KEY) > 50 else None
    }
    
    # Fallback to query parameter if it's a standard API key instead of a Read Only Bearer Token
    params = {"api_key": TMDB_API_KEY} if not headers["Authorization"] else {}

    # Step 1: Search for the TV series to get the ID
    search_url = f"{TMDB_BASE}/search/tv"
    search_params = {**params, "query": series_name}
    
    search_response = requests.get(search_url, headers=headers, params=search_params)
    search_response.raise_for_status()
    search_results = search_response.json().get("results", [])

    if not search_results:
        return f"No series found matching '{series_name}'."

    # Take the top match
    best_match = search_results[0]
    series_id = best_match["id"]
    official_name = best_match["name"]
    first_air_date = best_match.get("first_air_date", "Unknown date")

    # Step 2: Get watch providers for the series ID
    providers_url = f"{TMDB_BASE}/tv/{series_id}/watch/providers"
    providers_response = requests.get(providers_url, headers=headers, params=params)
    providers_response.raise_for_status()
    results = providers_response.json().get("results", {})

    # Filter by country
    country_data = results.get(country_code.upper())

    if not country_data:
        return f"'{official_name}' ({first_air_date[:4]}) found, but no streaming providers are listed for region '{country_code}'."

    # Parse different watch options
    streaming_options = {
        "Flatrate (Streaming)": [p["provider_name"] for p in country_data.get("flatrate", [])],
        "Buy": [p["provider_name"] for p in country_data.get("buy", [])],
        "Free": [p["provider_name"] for p in country_data.get("free", [])]
    }

    # Format the output
    output = f"Streaming Info for: **{official_name}** ({first_air_date[:4]})\n"
    output += f"Region: {country_code.upper()}\n"
    output += "---"
    
    has_providers = False
    for category, providers in streaming_options.items():
        if providers:
            has_providers = True
            output += f"\n* **{category}**: {', '.join(providers)}"
            
    if not has_providers:
        output += "\nAvailable to view, but no direct digital streaming/purchase platforms listed."

    return output


```
